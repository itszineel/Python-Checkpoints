```python
"""Multiply all elements of a list"""

size  = int(input("Enter the list size : "))
numbers = []

for i in range (size) :
    numbers.append(int(input("Enter your list : ")))

result = 1

for number in numbers:
    result = result * number

print("Result =", result)
```

```python
"""Sort tuples by the last element"""

tuples = [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]

tuples.sort(key=lambda x: x[-1])

print(tuples)
```

```python
"""Combine two dictionaries"""

d1 = {'a': 100, 'b': 200, 'c': 300}
d2 = {'a': 300, 'b': 200, 'd': 400}

result = d1.copy()

for key in d2:
    if key in result:
        result[key] = result[key] + d2[key]
    else:
        result[key] = d2[key]

print(result)
```

```python
"""Create a dictionary"""

n = int(input("Enter a number: "))

result = {}

for i in range(1, n + 1):
    result[i] = i * i

print(result)
```

```python
"""Sort tuples by their floating-point value"""

items = [
    ('item1', '12.20'),
    ('item2', '15.10'),
    ('item3', '24.5')
]

items.sort(key=lambda x: float(x[1]), reverse=True)

print(items)
```

```python
"""Create, iterate, add and remove elements from a set"""

numbers = {0, 1, 2, 3, 4}

print("Set:", numbers)

for number in numbers:
    print(number)

numbers.add(5)
numbers.add(6)

print("After adding elements:", numbers)

numbers.remove(2)

print("After removing an element:", numbers)
```
